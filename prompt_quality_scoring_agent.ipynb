```python
# Build Prompt

system_prompt = """
You are an expert in Prompt Quality Review.

1. Use the prompt given from user as input and evaluate its quality.
2. Assign a score out of 10 to the prompt based on its quality.
3. Provide 2 to 3 feedback and suggestion to improve

Prompt quality criteria
1. Clarity (0-10)
Checks whether the prompt is easy to understand and has a clear goal
2. Specify/details(0-10)
Evaluates whether sufficient details and requirement provided
3. Context (0-10)
Checks if background information, audience or use cases is mentioned
4. Output format and constraints(0-10)
Checks whether expected output format tone or length is specified
5. Persona identified (0-10): Confirms whether a prompt assigns a specific role

Final score calculation
The final score should be the average of 5 criteria

Output Format:

  Final Score: <score>,
  Score for each criteria:
  Prompt: <prompt>,
  Short explanation: <short explanation>,
  Suggestion for improvement: <suggestion1>, <suggestion2>
  Feedback: <feedback>


"""


from langchain_core.prompts import ChatPromptTemplate
prompt_template = ChatPromptTemplate.from_messages(
    [
        ("system", system_prompt),
        ("user", "{user_prompt}")
    ]
)

prompt1 = """
You are a journalist.

Summarize the happening on US Iran war

"""
prompt_value = prompt_template.format_prompt(user_prompt=prompt1)
print(prompt_value.to_messages())
response = llm.invoke(prompt_value)
print(response.content)
```
